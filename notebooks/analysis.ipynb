# 1. 프로젝트 설명
기업 재무·공시 데이터의 변화를 확인 후보로 분석한다. 실제 데이터와 분석 기준은 담당자가 확정한다.

## 2. 라이브러리
필요한 Python·Pandas 라이브러리와 버전을 기록한다.

## 3. 데이터 로드
실제 데이터 출처, 파일 경로, 수집일을 기록하고 로드 코드를 추가한다. 비밀 키는 기록하지 않는다.

## 4. 데이터 구조 확인
컬럼, 자료형, 기간, 기업 식별자를 확인한다.

## 5. 결측값/중복값 확인
검사 기준과 처리 근거를 기록한다.

## 6. 재무 분석
전기 대비 변화를 분석하고 계산식과 단위를 기록한다.

## 7. 공시 분석
기간별 공시 빈도와 유형을 분석한다.

## 8. SQL 결과와 비교
동일한 기업·기간·집계 기준에서 SQL과 Pandas 결과를 비교한다.

## 9. 검증 결과
일치 여부, 차이의 원인, 재검증 결과를 기록한다.

## 10. 결론
검증된 사실과 추가 검토할 확인 후보만 기록한다.
